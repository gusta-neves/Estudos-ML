## Dict

```python
dict = {
    'theo': 10,
    'leo': 8.2,
    1 : True,
    3.14 : [3, 1, 4, 1, 5],
    'livros' : [
        {"nome": 'John', "ano": 2025},
        {"nome": 'Alan', "ano": 2023}
        ]
}

print(dict.get('theo'))
print(dict['theo'])
# print(dict.get('ana)) -> None
print(dict.get(1))
print(dict[3.14][-1])
print(dict['livros'][1]["nome"])
print(dict.get('livros')[1].get("nome"))
```

```python
for i in dict:
    print(i, '->', dict[i])
```

```python
dict.items()
```

```python
for i, j in dict.items():
    print(i, '->', j)
```

## Function

```python
def f():
    print("hello")

f()
```

```python
# 'a : int' and '->int' this are just to say to python that the function expects to recieve an int and return int
# but they don't may to do.

# x = 0 it's just to say that when the user dont' pass the value, use 'x' as 0
def exp(a : int, x = 0) -> int:
    return a**x

print(exp(2,3))

print(exp(2.1, 3))

print(exp(2))
```

```python
def SumAndMul(a, b) -> int:
    return a + b, a * b

x, y = SumAndMul(10, 1)

print(f'Sum: {x}, Mul: {y}')

z = SumAndMul(10, 1)

print(z)
```

```python
def fibo(n : int) -> int:
    if (n == 0 or n == 1):
        return 1
    return (fibo(n-1) + fibo(n-2))


for i in range(8):
    print(f'fibonacci de {i} eh: {fibo(i)}')
```

```python
def f(x):
    return 2*x

print(f(2))
```

```python
ceps = [
    "12345678",
    "05140020",
    "04165080",
    "04236240",
    "02343010"
]
```

```python
url = "https://viacep.com.br/ws/{cep}/json/"

dados = []

for i in ceps:
    resposta = requests.get(url.format(cep = i))

    if resposta.status_code == 200:
        dados.append(resposta.json())

dados
type(dados)
dados[1]['cep']
```
